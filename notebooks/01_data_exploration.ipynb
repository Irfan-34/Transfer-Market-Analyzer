# ⚽ Transfer Market Exploration & Analytical Baseline

This notebook explores the historical transfer market dataset stored in `data/raw/transfermarkt.duckdb`.
We examine total transfer spend, season trends, league distribution, top transfers, position vs. transfer fee, and valuation correlations.

```python
import sys
from pathlib import Path
BASE_DIR = Path("..").resolve()
if str(BASE_DIR) not in sys.path:
    sys.path.append(str(BASE_DIR))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.data.db import query_duckdb
from src.analytics.market_analysis import (
    get_market_summary,
    get_spending_by_season,
    get_spending_by_league,
    get_top_transfers,
    get_position_fee_distribution
)
```

## 1. High-Level Transfer Market Summary

```python
summary = get_market_summary()
print("=== Transfer Market Macro Summary ===")
for k, v in summary.items():
    if isinstance(v, float):
        print(f"{k}: €{v:,.2f}")
    else:
        print(f"{k}: {v:,}")
```

## 2. Transfer Spending Trends by Season

```python
spending_season_df = pd.DataFrame(get_spending_by_season(limit=15))
display(spending_season_df)

plt.figure(figsize=(12, 5))
sns.barplot(data=spending_season_df, x="season", y="total_spending_eur", palette="viridis")
plt.title("Total Transfer Spending by Season (€)")
plt.xticks(rotation=45)
plt.ylabel("Total Spending (€)")
plt.tight_layout()
plt.show()
```

## 3. Spending by Domestic League

```python
league_df = pd.DataFrame(get_spending_by_league(limit=10))
display(league_df)

plt.figure(figsize=(10, 5))
sns.barplot(data=league_df, x="total_spending_eur", y="league_name", palette="rocket")
plt.title("Top 10 Leagues by Total Transfer Spending (€)")
plt.xlabel("Total Spending (€)")
plt.tight_layout()
plt.show()
```
